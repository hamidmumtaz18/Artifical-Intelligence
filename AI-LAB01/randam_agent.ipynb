```python

import random
import statistics

def run_agent(seed, agent_type, steps=20):
    rng = random.Random(seed)

    # Generate a random initial state
    rooms = {
        'A': rng.randint(0, 1),
        'B': rng.randint(0, 1)
    }
    location = rng.choice(['A', 'B'])

    score = 0

    for _ in range(steps):
        # Choose an action
        if agent_type == "Random":
            action = rng.choice(['Left', 'Right', 'Suck'])
        else:
            if rooms[location] == 1:
                action = 'Suck'
            elif location == 'A':
                action = 'Right'
            else:
                action = 'Left'

        # Execute the action and calculate score
        if action == 'Suck':
            if rooms[location] == 1:
                score += 10
            rooms[location] = 0

        elif action == 'Left':
            location = 'A'
            score -= 1

        elif action == 'Right':
            location = 'B'
            score -= 1

    return score


random_scores = []
reflex_scores = []

# Test on 10 random initial states, 20 steps per trial
for seed in range(10):
    random_scores.append(run_agent(seed, "Random", 20))
    reflex_scores.append(run_agent(seed, "Reflex", 20))

print("Random agent scores:", random_scores)
print("Reflex agent scores:", reflex_scores)

print("Random agent average:",
      statistics.mean(random_scores))

print("Reflex agent average:",
      statistics.mean(reflex_scores))
```
