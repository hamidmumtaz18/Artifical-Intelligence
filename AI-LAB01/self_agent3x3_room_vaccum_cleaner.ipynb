```python

import random

class VacuumEnvironment:
    """3x3 vacuum world with randomly distributed dirt."""

    def __init__(self):
        self.rooms = {
            (r, c): random.randint(0, 1)
            for r in range(3)
            for c in range(3)
        }
        self.agent_position = (0, 0)

    def percept(self):
        position = self.agent_position
        return position, self.rooms[position]

    def execute(self, action):
        r, c = self.agent_position

        if action == "Suck":
            self.rooms[(r, c)] = 0

        elif action == "Up":
            self.agent_position = (r - 1, c)

        elif action == "Down":
            self.agent_position = (r + 1, c)

        elif action == "Left":
            self.agent_position = (r, c - 1)

        elif action == "Right":
            self.agent_position = (r, c + 1)


class ModelBasedReflexAgent:
    """Maintains a model of rooms and visits all nine rooms."""

    def __init__(self):
        self.model = {}
        self.visited = set()
        self.path = [
            (0, 0), (0, 1), (0, 2),
            (1, 2), (1, 1), (1, 0),
            (2, 0), (2, 1), (2, 2)
        ]

    def act(self, percept):
        position, dirt = percept

        # Update the internal model
        self.model[position] = dirt
        self.visited.add(position)

        # Clean the current room if dirty
        if dirt == 1:
            return "Suck"

        # Find the next room in the planned route
        index = self.path.index(position)

        if index == len(self.path) - 1:
            return "Stop"

        next_position = self.path[index + 1]
        r, c = position
        nr, nc = next_position

        if nr < r:
            return "Up"
        elif nr > r:
            return "Down"
        elif nc < c:
            return "Left"
        else:
            return "Right"


# Run the simulation
random.seed(7)

env = VacuumEnvironment()
agent = ModelBasedReflexAgent()

moves = 0
cleaned_rooms = 0
score = 0

print("Initial Grid (1 = Dirty, 0 = Clean):")

for r in range(3):
    print([env.rooms[(r, c)] for c in range(3)])

print("\nAgent simulation:")

while True:
    percept = env.percept()
    position, dirt = percept
    action = agent.act(percept)

    if action == "Stop":
        break

    env.execute(action)

    if action == "Suck":
        cleaned_rooms += 1
        score += 10
    else:
        moves += 1
        score -= 1

    print(
        f"Position: {position}, Dirt: {dirt}, "
        f"Action: {action}"
    )

print("\nFinal Grid (0 = Clean):")

for r in range(3):
    print([env.rooms[(r, c)] for c in range(3)])

print("\n--- Performance Report ---")
print("Rooms visited:", len(agent.visited))
print("Moves:", moves)
print("Dirty rooms cleaned:", cleaned_rooms)
print("Performance score:", score)
print("All rooms clean:", all(v == 0 for v in env.rooms.values()))
```
