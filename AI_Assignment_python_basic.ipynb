```python

# ==========================================
# PYTHON FUNDAMENTALS
# Subject: Artificial Intelligence
# ==========================================

# 1. OUTPUT
print("Welcome to Python Programming!")

# 2. VARIABLES AND DATA TYPES
print("\n--- Variables and Data Types ---")

name = "Ali"
age = 21
marks = 85
is_student = True

print("Name:", name)
print("Age:", age)
print("Marks:", marks)
print("Student:", is_student)

# 3. USER INPUT
print("\n--- User Input ---")

user_name = input("Enter your name: ")
user_age = int(input("Enter your age: "))

print("Hello", user_name)
print("Your age is", user_age)

# 4. OPERATORS
print("\n--- Arithmetic Operators ---")

a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)

# 5. CONDITIONAL STATEMENTS
print("\n--- Conditions ---")

marks = 75

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
else:
    print("Grade C")

# 6. LOOPS
print("\n--- For Loop ---")

for i in range(1, 6):
    print(i)

print("\n--- While Loop ---")

counter = 1

while counter <= 5:
    print(counter)
    counter += 1

# 7. LISTS
print("\n--- Lists ---")

fruits = ["Apple", "Banana", "Mango"]

fruits.append("Orange")

print("Fruits:", fruits)
print("First Fruit:", fruits[0])

# 8. DICTIONARIES
print("\n--- Dictionaries ---")

student = {
    "name": "Ali",
    "age": 21,
    "department": "BSCS"
}

print(student)
print("Student Name:", student["name"])

# 9. FUNCTIONS
print("\n--- Functions ---")

def add_numbers(a, b):
    return a + b

print("Sum:", add_numbers(10, 20))

# FINAL MESSAGE
print("\nPython Fundamentals Completed!")



```
