```python

from collections import deque
import time

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

MOVES = {
    'Up': -3,
    'Down': 3,
    'Left': -1,
    'Right': 1
}


def actions(state):
    i = state.index(0)
    row, col = divmod(i, 3)
    legal = []

    if row > 0:
        legal.append('Up')
    if row < 2:
        legal.append('Down')
    if col > 0:
        legal.append('Left')
    if col < 2:
        legal.append('Right')

    return legal


def result(state, action):
    i = state.index(0)
    j = i + MOVES[action]

    s = list(state)
    s[i], s[j] = s[j], s[i]

    return tuple(s)


def is_solvable(state):
    tiles = [x for x in state if x != 0]

    inversions = sum(
        1
        for i in range(len(tiles))
        for j in range(i + 1, len(tiles))
        if tiles[i] > tiles[j]
    )

    return inversions % 2 == 0


def show(state):
    for r in range(3):
        print(' '.join(
            str(x) if x != 0 else '_'
            for x in state[r * 3:r * 3 + 3]
        ))


def solve_bfs(start):
    frontier = deque([start])
    parent = {start: (None, None)}
    expanded = 0

    while frontier:
        state = frontier.popleft()

        if state == GOAL:
            path = []
            current = state

            while parent[current][0] is not None:
                previous, action = parent[current]
                path.append(action)
                current = previous

            path.reverse()
            return path, expanded, len(parent)

        expanded += 1

        for action in actions(state):
            child = result(state, action)

            if child not in parent:
                parent[child] = (state, action)
                frontier.append(child)

    return None, expanded, len(parent)


# Given start state
start = (8, 1, 3,
         4, 0, 2,
         7, 6, 5)

print("Start state:")
show(start)

if not is_solvable(start):
    print("This puzzle is unsolvable.")
else:
    t0 = time.perf_counter()

    path, expanded, generated = solve_bfs(start)

    t1 = time.perf_counter()

    print("\nSolvable:", True)

    if path is not None:
        print("Solution:", path)
        print("Moves:", len(path))
        print("Nodes expanded:", expanded)
        print("Nodes generated:", generated)
        print(f"Time taken: {(t1 - t0) * 1000:.2f} ms")
    else:
        print("No solution found.")
```
