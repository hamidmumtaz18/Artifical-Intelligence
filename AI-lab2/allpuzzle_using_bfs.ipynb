```python
from collections import deque
import time

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)          # 0 represents the blank tile

MOVES = {'Up': -3, 'Down': 3, 'Left': -1, 'Right': 1}

def actions(state):
    """Return the legal moves of the blank tile."""
    i = state.index(0)
    row, col = divmod(i, 3)
    legal = []
    if row > 0: legal.append('Up')
    if row < 2: legal.append('Down')
    if col > 0: legal.append('Left')
    if col < 2: legal.append('Right')
    return legal

def result(state, action):
    """Transition model: swap the blank with the neighbouring tile."""
    i = state.index(0)
    j = i + MOVES[action]
    s = list(state)
    s[i], s[j] = s[j], s[i]
    return tuple(s)

def goal_test(state):
    return state == GOAL

def is_solvable(state):
    tiles = [t for t in state if t != 0]
    inversions = sum(1 for a in range(len(tiles))
                       for b in range(a + 1, len(tiles)) if tiles[a] > tiles[b])
    return inversions % 2 == 0

def solve_bfs(start):
    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0
    while frontier:
        state, path = frontier.popleft()
        if goal_test(state):
            return path, expanded, len(explored)
        expanded += 1
        for a in actions(state):
            child = result(state, a)
            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [a]))
    return None, expanded, len(explored)

def show(state):
    for r in range(3):
        print(' '.join(str(x) if x else '_' for x in state[r*3:r*3+3]))

# --- UPDATED START STATE ---
start = (8, 1, 4,
         0, 2, 7,
         6, 5, 3)

print("Start state:"); show(start)
print("Solvable:", is_solvable(start))
t0 = time.perf_counter()
path, expanded, generated = solve_bfs(start)
t1 = time.perf_counter()

print("\n--- Results ---")
print("Solution :", path)
print("Moves    :", len(path) if path else "No solution")
print("Nodes expanded :", expanded)
print("Nodes generated:", generated)
print(f"Time taken     : {(t1-t0)*1000:.2f} ms")
```

```python
from collections import deque
import time

# Goal state
GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)


# Check if the goal is reached
def goal_test(state):
    return state == GOAL


# Find possible moves
def actions(state):
    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = []

    if row > 0:
        moves.append("Up")
    if row < 2:
        moves.append("Down")
    if col > 0:
        moves.append("Left")
    if col < 2:
        moves.append("Right")

    return moves


# Apply a move and return the new state
def result(state, action):
    zero = state.index(0)

    if action == "Up":
        swap = zero - 3
    elif action == "Down":
        swap = zero + 3
    elif action == "Left":
        swap = zero - 1
    else:
        swap = zero + 1

    new_state = list(state)
    new_state[zero], new_state[swap] = (
        new_state[swap], new_state[zero]
    )

    return tuple(new_state)


# BFS algorithm
def solve_bfs(start):
    frontier = deque([(start, [])])
    explored = {start}
    expanded = 0

    while frontier:
        state, path = frontier.popleft()

        if goal_test(state):
            return path, expanded, len(explored)

        expanded += 1

        for action in actions(state):
            child = result(state, action)

            if child not in explored:
                explored.add(child)
                frontier.append((child, path + [action]))

    return None, expanded, len(explored)


# Print the board in 3x3 format
def print_board(state):
    for i in range(0, 9, 3):
        row = state[i:i + 3]
        print(*row)
    print()


# Print the board after every move
def print_solution(start, path):
    state = start

    print("Initial State:")
    print_board(state)

    for step, move in enumerate(path, start=1):
        state = result(state, move)

        print("Move", step, ":", move)
        print_board(state)

    print("Goal reached!")


# Main program
start = (8, 1, 3, 4, 0, 2, 7, 6, 5)

begin = time.perf_counter()

path, expanded, generated = solve_bfs(start)

end = time.perf_counter()

# Display results
if path is not None:
    print("Solution Length:", len(path))
    print("Nodes Expanded:", expanded)
    print("Nodes Generated:", generated)
    print("Execution Time:", end - begin, "seconds")
    print("\nSolution Boards:\n")

    print_solution(start, path)
else:
    print("No solution found.")
```
